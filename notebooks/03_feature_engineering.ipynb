```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/materials_clean.csv")
```

```python
# Band gap transformations

# Log transformation (avoid log(0))
df["band_gap_log"] = np.log1p(df["band_gap"])

# Categorize materials
def classify_material(bg):
    if bg < 0.1:
        return "metal"
    elif bg < 2:
        return "semiconductor"
    else:
        return "insulator"

df["material_type"] = df["band_gap"].apply(classify_material)
```

```python
# Energy-related features

# Stability indicator
df["is_stable"] = df["energy_above_hull"] < 0

# Normalize formation energy
df["formation_energy_abs"] = df["formation_energy_per_atom"].abs()
```

```python
# Density-based features

# Log density
df["density_log"] = np.log1p(df["density"])

# Density categories
df["density_category"] = pd.qcut(df["density"], q=3, labels=["low", "medium", "high"])
```

```python
# Interaction features

df["density_bandgap_ratio"] = df["density"] / (df["band_gap"] + 1e-6)
df["energy_density_product"] = df["formation_energy_per_atom"] * df["density"]
```

```python
# Metal validation feature

df["metal_consistency"] = df["is_metal"] == (df["band_gap"] < 0.1)
```

```python
# Encoding categorical features

df = pd.get_dummies(df, columns=["material_type", "density_category"], drop_first=True)
```

```python
# Save feature-engineered dataset

df.to_csv("../data/processed/materials_features.csv", index=False)
```

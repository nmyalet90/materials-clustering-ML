```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()

df = pd.read_csv("../data/processed/materials_features.csv")
```

```python
## Band gap distribution

plt.hist(df["band_gap"], bins=50)
plt.title("Band Gap Distribution")
plt.xlabel("Band Gap (eV)")
plt.ylabel("Frequency")
plt.show()
```

```python
## Log-transformed band gap

# Log transformation reduces skewness

plt.hist(df["band_gap_log"], bins=50)
plt.title("Log Band Gap Distribution")
plt.xlabel("log(1 + band gap)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Density distribution

plt.hist(df["density"], bins=50)
plt.title("Density Distribution")
plt.xlabel("Density")
plt.ylabel("Frequency")
plt.show()
```

```python
# Band gap vs density

plt.scatter(df["density"], df["band_gap"], alpha=0.5)
plt.xlabel("Density")
plt.ylabel("Band Gap")
plt.title("Density vs Band Gap")
plt.show()
```

```python
# Formation energy vs stability

sns.boxplot(x=df["is_stable"], y=df["formation_energy_per_atom"])
plt.title("Formation Energy vs Stability")
plt.show()
```

```python
# Material type distribution

df["material_type_metal"].sum(), df["material_type_semiconductor"].sum(), df["material_type_insulator"].sum()
```

```python
# Metal consistency check

df["metal_consistency"].value_counts()
```

```python
# Correlation matrix

corr = df.select_dtypes(include=["float64", "int64"]).corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=False, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
## Key Insights

# - Band gap distribution is highly skewed, justifying log transformation  
# - Stable materials tend to have lower formation energy  
# - Density and band gap show weak/moderate correlation  
# - Engineered features help reveal physical relationships  
```
